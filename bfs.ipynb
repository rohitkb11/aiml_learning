```python
import time
import queue

maze=[
    "S.....",
    ".####.",
    "......",
    ".####.",
    "E....."
]

for r in range(len(maze)):
    for c in range(len(maze[0])):
        if(maze[r][c]=='S'):
            start=(r,c)
        elif(maze[r][c]=='E'):
            end=(r,c)

print("Maze start: ",start)
print("Maze end: ",end)
        
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'F'],
    'E': ['C', 'F'],
    'F': ['D', 'E']
}

graph_start = 'A'
graph_end = 'F'
```

```python
mode="maze"
def get_neighbours(node):
    if mode=="graph":
        return graph[node]

    r,c=node

    dx=[-1,1,0,0]
    dy=[0,0,1,-1]

    results=[]

    for i in range(4):
        newR=r+dx[i]
        newC=c+dy[i]
        if 0<=newR<len(maze) and 0<=newC<len(maze[0]):
            if(maze[newR][newC]!="#"):
                results.append((newR,newC))

    return results
```

```python
def bfs(start,end):
    queue= [[start]]
    visited= [start]

    explored=0


    while len(queue)>0:
        path=queue.pop(0)
        node=path[-1]
        explored=explored+1

        if(node==end):
            return path,explored

        for n in get_neighbours(node):
            if n not in visited:
                queue.append(path+[n])
                visited.append(n)

    return None,explored
```

```python
def dfs(start,end):
    stack= [[start]]
    visited= [start]

    explored=0


    while len(stack)>0:
        path=stack.pop()
        node=path[-1]
        explored=explored+1

        if(node==end):
            return path,explored

        for n in get_neighbours(node):
            if n not in visited:
                stack.append(path+[n])
                visited.append(n)

    return None,explored
```

```python
def comparision(start,end):
    t1=time.perf_counter()
    bfs_path,bfs_nodes=bfs(start,end)
    t2=time.perf_counter()-t1
    
    t3=time.perf_counter()
    dfs_path,dfs_nodes=dfs(start,end)
    t4=time.perf_counter()-t3

    print("--------BFS----------")
    print("Path: ",bfs_path)
    print("Nodes: ",bfs_nodes)
    print("Time: ",t2)
    
    print("--------DFS----------")
    print("Path: ",dfs_path)
    print("Nodes: ",dfs_nodes)
    print("Time: ",t4)

    return  bfs_path, bfs_nodes, dfs_path, dfs_nodes  # ✅ Correct
```

```python
def show_path(path):
    if path is None:
        print("No path found")
        return

    grid=[list(row) for row in maze]

    for r,c in path:
        if(grid[r][c]=="."):
            grid[r][c]="*"

    for row in grid:
        print("".join(row))
```

```python
mode = "maze"

bfs_path, bfs_nodes, dfs_path, dfs_nodes = comparision(start, end)

print("\nBFS path:")
show_path(bfs_path)

print("\nDFS path:")
show_path(dfs_path)
```

```python
mode = "graph"

bfs_path, bfs_nodes, dfs_path, dfs_nodes = comparision(graph_start, graph_end)

print("\nBFS path:")
print(bfs_path)

print("\nDFS path:")
dfs_path
```
