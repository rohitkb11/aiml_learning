```python
class CollegeEnvironment:
    def __init__(self):
        self.seats = {"Computer": 120, "IT": 120}
        self.placement = {"Computer": 95, "IT": 90}
```

```python
class Sensors:
    def get_input(self):
        return input("Enter your choice: ")
```

```python
class Actuators:
    def show_courses(self):
        print("Courses: Computer Engineering, IT Engineering")

    def show_seats(self, environment):
        print("Vacant seats:", environment.seats)

    def confirm_admission(self, course, environment):
        if environment.seats[course] > 0:
            environment.seats[course] -= 1
            print("Admission confirmed in", course)
            print("Remaining seats:", environment.seats)
        else:
            print("No seats available in", course)
```

```python
class SimpleReflexAgent:
    def __init__(self, environment, actuators):
        self.environment = environment
        self.actuators = actuators

    def act(self, choice):
        if choice == "1":
            self.actuators.show_courses()

        elif choice == "2":
            self.actuators.show_seats(self.environment)

        elif choice == "3":
            print("Admission confirmed in Computer")

        elif choice == "4":
            print("Admission confirmed in IT")

        elif choice == "5":
            print("Recommended: Computer")

        else:
            print("Invalid choice")
```

```python
class ModelBasedAgent:
    def __init__(self, environment, actuators):
        self.environment = environment
        self.actuators = actuators

    def admission(self, course):
        self.actuators.confirm_admission(
            course, self.environment
        )

    def recommend_course(self):
        seats = self.environment.seats
        placement = self.environment.placement

        if seats["Computer"] == 0 and seats["IT"] == 0:
            print("No seats available")

        elif seats["Computer"] == 0:
            print("Recommended: IT")

        elif seats["IT"] == 0:
            print("Recommended: Computer")

        else:
            score_comp = (
                seats["Computer"] * 0.3
                + placement["Computer"] * 0.7
            )

            score_it = (
                seats["IT"] * 0.3
                + placement["IT"] * 0.7
            )

            if score_comp >= score_it:
                print("Recommended: Computer")
            else:
                print("Recommended: IT")

    def act(self, choice):
        if choice == "1":
            self.actuators.show_courses()

        elif choice == "2":
            self.actuators.show_seats(self.environment)

        elif choice == "3":
            self.admission("Computer")

        elif choice == "4":
            self.admission("IT")

        elif choice == "5":
            self.recommend_course()

        else:
            print("Invalid choice")
```

```python
print("COLLEGE ADMISSION CHATBOT")

print("\nPEAS COMPONENTS")
print("Performance Measure: Correct admission and recommendations")
print("Environment: College courses, seats and placement records")
print("Actuators: Display information and process admission")
print("Sensors: Take user input")

print("\nChoose Agent")
print("1. Simple Reflex Agent")
print("2. Model-Based Agent")

agent_choice = input("Enter your choice: ")

environment = CollegeEnvironment()
sensors = Sensors()
actuators = Actuators()

if agent_choice == "1":
    agent = SimpleReflexAgent(environment, actuators)

elif agent_choice == "2":
    agent = ModelBasedAgent(environment, actuators)

else:
    print("Invalid choice")
    exit()

while True:
    print("\n----- MENU -----")
    print("1. View courses")
    print("2. View vacant seats")
    print("3. Computer Engineering admission")
    print("4. IT Engineering admission")
    print("5. Recommend course")
    print("0. Exit")

    choice = sensors.get_input()

    if choice == "0":
        print("Thank you!")
        break

    agent.act(choice)
```
